```python
from ultralytics import YOLO
import torch
import os
import shutil
from pathlib import Path
import random

DRIVE_PATH = '/content/drive/MyDrive/CariesProject'

print("="*60)
print(" БЫСТРОЕ ОБУЧЕНИЕ НА РУЧНОЙ РАЗМЕТКЕ")
print("="*60)
print("="*60 + "\n")

# Создаём датасет
os.makedirs('/content/quick_dataset/train/images', exist_ok=True)
os.makedirs('/content/quick_dataset/train/labels', exist_ok=True)
os.makedirs('/content/quick_dataset/val/images', exist_ok=True)
os.makedirs('/content/quick_dataset/val/labels', exist_ok=True)

rf_images = '/content/roboflow_data/train/images'
rf_labels = '/content/roboflow_data/train/labels'

# Проверка
if not os.path.exists(rf_images):
    raise FileNotFoundError(f" Папка не найдена: {rf_images}\nЗапустите ЯЧЕЙКУ 5!")

# Сбор пар
pairs = []
for img_file in os.listdir(rf_images):
    if img_file.endswith(('.jpg', '.png', '.jpeg')):
        lbl_file = Path(img_file).stem + '.txt'
        if os.path.exists(f'{rf_labels}/{lbl_file}'):
            pairs.append((img_file, lbl_file))

if len(pairs) == 0:
    raise ValueError(" Не найдено пар изображение+разметка!")

print(f" Найдено {len(pairs)} размеченных изображений\n")

# Делим 80/20
random.seed(42)
random.shuffle(pairs)
split = int(len(pairs) * 0.8)
train_pairs = pairs[:split]
val_pairs = pairs[split:]

# Копируем
for img, lbl in train_pairs:
    shutil.copy2(f'{rf_images}/{img}', f'/content/quick_dataset/train/images/{img}')
    shutil.copy2(f'{rf_labels}/{lbl}', f'/content/quick_dataset/train/labels/{lbl}')

for img, lbl in val_pairs:
    shutil.copy2(f'{rf_images}/{img}', f'/content/quick_dataset/val/images/{img}')
    shutil.copy2(f'{rf_labels}/{lbl}', f'/content/quick_dataset/val/labels/{lbl}')

# YAML
yaml_content = """path: /content/quick_dataset
train: train/images
val: val/images
nc: 1
names: ['caries']
"""

with open('/content/quick_dataset/data.yaml', 'w') as f:
    f.write(yaml_content)

print(f" Датасет: {len(train_pairs)} train, {len(val_pairs)} val\n")

# Обучение
model = YOLO('yolov8m.pt')

print(" Начало обучения...")
print("="*60 + "\n")

results = model.train(
    data='/content/quick_dataset/data.yaml',
    epochs=30,
    imgsz=640,
    batch=16,
    device=0,
    patience=10,
    save=True,
    project=f'{DRIVE_PATH}/runs',
    name='quick_train',
    exist_ok=True,
    verbose=True,
)

print("\n" + "="*60)
print("ОБУЧЕНИЕ ЗАВЕРШЕНО!")
print("="*60)

# Сохранение
best_path = f'{DRIVE_PATH}/runs/quick_train/weights/best.pt'
os.makedirs(f'{DRIVE_PATH}/models', exist_ok=True)
shutil.copy2(best_path, f'{DRIVE_PATH}/models/quick_model.pt')

print(f"\n Модель сохранена:")
print(f"   {best_path}")
print(f"   {DRIVE_PATH}/models/quick_model.pt")
print("\n Готово к авто-разметке!")
```

```python
from ultralytics import YOLO
import os
import shutil
from pathlib import Path
from tqdm import tqdm
import random

DRIVE_PATH = '/content/drive/MyDrive/CariesProject'

# ========== НАСТРОЙКИ ==========
MODEL_PATH = f'{DRIVE_PATH}/models/quick_model.pt'
AUTO_LABEL_DIR = '/content/auto_labeled'
RAW_DATASETS = '/content/raw_datasets'

CONF_THRESHOLD = 0.30    # Порог уверенности (ниже = больше объектов)
IOU_THRESHOLD = 0.45
MAX_IMAGES = 700        # Ограничение для скорости

print("="*60)
print(" АВТО-РАЗМЕТКА ДАТАСЕТОВ")
print("="*60)
print(f"Модель: {MODEL_PATH}")
print(f"Порог уверенности: {CONF_THRESHOLD}")
print(f"Макс. изображений: {MAX_IMAGES}")
print("="*60 + "\n")

# Загрузка модели
if not os.path.exists(MODEL_PATH):
    raise FileNotFoundError(f" Модель не найдена: {MODEL_PATH}")

print("Загрузка модели...")
model = YOLO(MODEL_PATH)
print(" Модель загружена\n")

# Создание структуры
os.makedirs(f'{AUTO_LABEL_DIR}/images', exist_ok=True)
os.makedirs(f'{AUTO_LABEL_DIR}/labels', exist_ok=True)

# Сбор изображений
print(" Поиск изображений в датасетах...")

all_images = []

for ds_folder in ['ds1', 'ds2', 'ds3']:
    ds_path = f'{RAW_DATASETS}/{ds_folder}'

    if not os.path.exists(ds_path):
        continue

    for root, dirs, files in os.walk(ds_path):
        for file in files:
            if file.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp')):
                full_path = os.path.join(root, file)
                all_images.append((full_path, ds_folder))

print(f" Найдено: {len(all_images)} изображений")

# Ограничиваем
if len(all_images) > MAX_IMAGES:
    random.seed(42)
    all_images = random.sample(all_images, MAX_IMAGES)
    print(f"  Ограничено до: {MAX_IMAGES} изображений\n")

# Авто-разметка
print("  Начинаю авто-разметку...")
print("-" * 60)

labeled_count = 0
skipped_count = 0
total_boxes = 0

for img_path, ds_name in tqdm(all_images, desc="Обработка"):
    try:
        # Предсказание
        results = model.predict(
            source=img_path,
            conf=CONF_THRESHOLD,
            iou=IOU_THRESHOLD,
            verbose=False,
            device=0
        )

        # Пропускаем пустые
        if len(results[0].boxes) == 0:
            skipped_count += 1
            continue

        # Уникальное имя
        orig_name = Path(img_path).stem
        ext = Path(img_path).suffix
        unique_name = f"{ds_name}_{orig_name}"

        # Копируем изображение
        dst_img = f'{AUTO_LABEL_DIR}/images/{unique_name}{ext}'
        shutil.copy2(img_path, dst_img)

        # Создаём разметку
        dst_lbl = f'{AUTO_LABEL_DIR}/labels/{unique_name}.txt'

        with open(dst_lbl, 'w') as f:
            for box in results[0].boxes:
                x, y, w, h = box.xywhn[0].cpu().numpy()
                cls = int(box.cls[0])

                f.write(f"{cls} {x:.6f} {y:.6f} {w:.6f} {h:.6f}\n")
                total_boxes += 1

        labeled_count += 1

    except Exception as e:
        skipped_count += 1

# Статистика
print("\n" + "="*60)
print(" РЕЗУЛЬТАТЫ АВТО-РАЗМЕТКИ")
print("="*60)
print(f"Обработано:              {len(all_images)}")
print(f"Размечено (с объектами): {labeled_count}")
print(f"Пропущено (пусто):       {skipped_count}")
print(f"Всего объектов:          {total_boxes}")
print(f"Среднее на изображение:  {total_boxes / labeled_count if labeled_count > 0 else 0:.1f}")
print("="*60)

# Сохранение на Drive
print("\n Сохранение на Google Drive...")
drive_auto = f'{DRIVE_PATH}/auto_labeled'
if os.path.exists(drive_auto):
    shutil.rmtree(drive_auto)
shutil.copytree(AUTO_LABEL_DIR, drive_auto)
print(" Сохранено!\n")

print("="*60)
print(" АВТО-РАЗМЕТКА ЗАВЕРШЕНА")
print("="*60)
```

```python
import os
import shutil
import random
from pathlib import Path
from tqdm import tqdm

DRIVE_PATH = '/content/drive/MyDrive/CariesProject'

# Источники
ROBOFLOW_DATA = '/content/roboflow_data/train'
AUTO_LABELED = '/content/auto_labeled'

FINAL_DATASET = '/content/dataset_complete'

TRAIN_RATIO = 0.85
random.seed(42)

print("="*60)
print(" ОБЪЕДИНЕНИЕ ВСЕХ ДАННЫХ")
print("="*60)

# Структура
for subset in ['train', 'val']:
    for folder in ['images', 'labels']:
        os.makedirs(f'{FINAL_DATASET}/{subset}/{folder}', exist_ok=True)

# Сбор всех пар
all_pairs = []

# 1. Ручная разметка (Roboflow) - высокий приоритет
print("\n 1/2: Сбор ручной разметки...")
rf_count = 0

if os.path.exists(ROBOFLOW_DATA):
    rf_images = f'{ROBOFLOW_DATA}/images'
    rf_labels = f'{ROBOFLOW_DATA}/labels'

    for img_file in os.listdir(rf_images):
        if not img_file.lower().endswith(('.jpg', '.jpeg', '.png')):
            continue

        lbl_file = Path(img_file).stem + '.txt'

        img_path = f'{rf_images}/{img_file}'
        lbl_path = f'{rf_labels}/{lbl_file}'

        if os.path.exists(lbl_path):
            all_pairs.append({
                'img': img_path,
                'lbl': lbl_path,
                'name': f'manual_{img_file}',
                'source': 'roboflow',
                'priority': 1
            })
            rf_count += 1

print(f"    Найдено: {rf_count}")

# 2. Авто-разметка
print("\n 2/2: Сбор авто-разметки...")
auto_count = 0

if os.path.exists(AUTO_LABELED):
    auto_images = f'{AUTO_LABELED}/images'
    auto_labels = f'{AUTO_LABELED}/labels'

    for img_file in os.listdir(auto_images):
        if not img_file.lower().endswith(('.jpg', '.jpeg', '.png')):
            continue

        lbl_file = Path(img_file).stem + '.txt'

        img_path = f'{auto_images}/{img_file}'
        lbl_path = f'{auto_labels}/{lbl_file}'

        if os.path.exists(lbl_path):
            all_pairs.append({
                'img': img_path,
                'lbl': lbl_path,
                'name': f'auto_{img_file}',
                'source': 'auto',
                'priority': 2
            })
            auto_count += 1

print(f"    Найдено: {auto_count}")

# Статистика
total = len(all_pairs)

print(f"\n{'='*60}")
print(f" ВСЕГО ДАННЫХ")
print(f"{'='*60}")
print(f"Ручная разметка:  {rf_count:4d} ({rf_count/total*100:.1f}%)")
print(f"Авто-разметка:    {auto_count:4d} ({auto_count/total*100:.1f}%)")
print(f"{'─'*60}")
print(f"ИТОГО:            {total:4d}")
print(f"{'='*60}")

# Разделение
print(f"\n Разделение на train/val...")

manual_pairs = [p for p in all_pairs if p['source'] == 'roboflow']
auto_pairs = [p for p in all_pairs if p['source'] == 'auto']

random.shuffle(auto_pairs)

# Все ручные - в train
train_pairs = manual_pairs.copy()

# Дополняем train
remaining = int(len(auto_pairs) * TRAIN_RATIO)
train_pairs.extend(auto_pairs[:remaining])

# Остальное в val
val_pairs = auto_pairs[remaining:]

print(f"   Train: {len(train_pairs)}")
print(f"   Val:   {len(val_pairs)}")

# Копирование
print(f"\n Копирование файлов...")

for subset, pairs in [('train', train_pairs), ('val', val_pairs)]:
    for pair in tqdm(pairs, desc=f"  {subset}"):
        ext = Path(pair['img']).suffix

        dst_img = f"{FINAL_DATASET}/{subset}/images/{pair['name']}"
        dst_lbl = f"{FINAL_DATASET}/{subset}/labels/{Path(pair['name']).stem}.txt"

        shutil.copy2(pair['img'], dst_img)
        shutil.copy2(pair['lbl'], dst_lbl)

# YAML
yaml_content = f"""path: {FINAL_DATASET}
train: train/images
val: val/images

nc: 1
names: ['caries']

# Dataset composition
# Manual: {rf_count}
# Auto: {auto_count}
# Total: {total}
# Train: {len(train_pairs)}
# Val: {len(val_pairs)}
"""

with open(f'{FINAL_DATASET}/data.yaml', 'w') as f:
    f.write(yaml_content)

# Сохранение на Drive
print(f"\n Сохранение на Google Drive...")

drive_final = f'{DRIVE_PATH}/dataset_complete'
if os.path.exists(drive_final):
    shutil.rmtree(drive_final)

shutil.copytree(FINAL_DATASET, drive_final)

print(f"\n{'='*60}")
print(f" ФИНАЛЬНЫЙ ДАТАСЕТ ГОТОВ")
print(f"{'='*60}")
print(f"Путь: {FINAL_DATASET}")
print(f"Train: {len(train_pairs)} | Val: {len(val_pairs)}")
print(f"{'='*60}")
```

```python
from ultralytics import YOLO
import torch
import os
import shutil

assert torch.cuda.is_available(), " GPU не найдена!"

DRIVE_PATH = '/content/drive/MyDrive/CariesProject'
DATASET = '/content/dataset_complete/data.yaml'

print("="*60)
print(" ФИНАЛЬНОЕ ОБУЧЕНИЕ")
print("="*60)
print(f"GPU: {torch.cuda.get_device_name(0)}")
print("="*60 + "\n")

# Продолжить обучение
model = YOLO(f'{DRIVE_PATH}/models/quick_model.pt')

print(" Начало финального обучения...")
print("="*60 + "\n")

results = model.train(
    data=DATASET,

    epochs=50,           # Больше эпох
    imgsz=1280,          # Больше разрешение
    batch=4,
    device=0,

    optimizer='AdamW',
    lr0=0.0005,          # Низкий LR для fine-tuning
    lrf=0.00001,
    warmup_epochs=3,
    cos_lr=True,

    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    degrees=10.0,
    translate=0.15,
    scale=0.5,
    shear=3.0,
    flipud=0.0,
    fliplr=0.5,
    mosaic=1.0,
    mixup=0.2,
    copy_paste=0.2,
    erasing=0.4,
    auto_augment='randaugment',

    box=7.5,
    cls=0.5,
    dfl=1.5,

    patience=20,
    save=True,
    plots=True,

    project=f'{DRIVE_PATH}/runs',
    name='caries_FINAL',
    exist_ok=True,
    verbose=True,
)

print("\n ОБУЧЕНИЕ ЗАВЕРШЕНО!")

# Валидация
metrics = model.val(data=DATASET, imgsz=1280)

map50 = float(metrics.box.map50)
recall = float(metrics.box.r)
precision = float(metrics.box.p)

print(f"""
╔════════════════════════════════════╗
║      ФИНАЛЬНЫЕ МЕТРИКИ          ║
╠════════════════════════════════════╣
║  mAP@0.5:    {map50:.4f}              ║
║  Precision:  {precision:.4f}              ║
║  Recall:     {recall:.4f}              ║
╚════════════════════════════════════╝
""")

# Сохранение
best_path = f'{DRIVE_PATH}/runs/caries_FINAL/weights/best.pt'
os.makedirs(f'{DRIVE_PATH}/models', exist_ok=True)
shutil.copy2(best_path, f'{DRIVE_PATH}/models/final_model.pt')

print(f" Модель: {DRIVE_PATH}/models/final_model.pt")
```
